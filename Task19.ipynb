```python
#Duplicate Record Validation
df.duplicated().sum()
```

```python
#Display duplicate rows
df[df.duplicated()]
```

```python
#Remove duplicate rows
df_cleaned = df.drop_duplicates()
```

```python
#Verify duplicates were removed
df_cleaned.duplicated().sum()
```

```python
#Data Validation Checks
#Missing Values
df_cleaned.isnull().sum()
```

```python
#Duplicate Customer IDs
df_cleaned["Customer Id"].duplicated().sum()
```

```python
df_cleaned[df_cleaned["Customer Id"].duplicated(keep=False)]
```

```python
#Email Validation
invalid_email = df_cleaned[
    ~df_cleaned["Email"].astype(str).str.contains("@", na=False)
]

invalid_email
```

```python
#Phone Number Validation
phone_digits = (
    df_cleaned["Phone 1"]
    .astype(str)
    .str.replace(r"\D", "", regex=True)
)

invalid_phone = df_cleaned[
    (phone_digits.str.len() < 7) |
    (phone_digits.str.len() > 15)
]

invalid_phone
```

```python
#Website Validation
invalid_website = df_cleaned[
    ~df_cleaned["Website"].astype(str).str.contains(
        r"\.",
        regex=True,
        na=False
    )
]

invalid_website
```

```python
#First Name Validation
df_cleaned["First Name"].isnull().sum()
```

```python
#Country Validation
df_cleaned["Country"].nunique()
```

```python
#Final Dataset Checks
#Check missing values again
df_cleaned.isnull().sum()
```

```python
#Check duplicates again
df_cleaned.duplicated().sum()
```

```python
#Check Customer ID uniqueness
df_cleaned["Customer Id"].is_unique
```

```python
#View the cleaned data
df_cleaned.head()
```

```python
#Save Cleaned Dataset
df_cleaned.to_csv(
    "cleaned_customer_dataset.csv",
    index=False
)
```
